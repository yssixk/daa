```python
from pulp import LpProblem, LpVariable, LpMinimize, LpStatus

model = LpProblem("Pemecahan_masalah", sense=LpMinimize)
x = LpVariable("x", lowBound=0)
y = LpVariable("y", lowBound=0)
model += x + y, "Obj"
model += 4 * x + 3 * y == 34, "kendala_2"
model.solve()
print("status:", LpStatus[model.status])
print("x =", x.value())
print("y =", y.value())
```

```python
from sympy import symbols, Eq, solve

P, L = symbols('P L')
eq1 = Eq(2 * (P + L), 44)
eq2 = Eq(L, P - 6)
solusi = solve((eq1, eq2), (P, L))
panjang = solusi[P]
lebar = solusi[L]
print("Panjang =", panjang, "cm")
print("Lebar =", lebar, "cm")
```

```python
from scipy.optimize import linprog

A = [[4, 3],
     [3, 2],
     [-1, 0],
     [0, -1]]

b = [20, 30, 17, 0]
C = [-5, -7]

res = linprog(C, A_ub=A, b_ub=b)
```

```python
print('Optimal value:', round(res.fun * -1, ndigits=2),
      '\nX values:', res.x,
      '\nNumber of iterations performed:', res.nit,
      '\nStatus:', res.message)
```

```python
from scipy.optimize import linprog

C = [1, 1]  # Minimize x + y
A_eq = [[4, 12]]
b_eq = [28]

res = linprog(C, A_eq=A_eq, b_eq=b_eq, method='highs')

print("x =", res.x[0])
print("y =", res.x[1])
print("Status:", res.message)
```

```python
from scipy.optimize import linprog
C = [0, 0]
A_eq = [[1, 1],
        [3, 4]]
b_eq = [16, 55]
bounds = [(0, None), (0, None)]
res = linprog(C, A_eq=A_eq, b_eq=b_eq, bounds=bounds, method='highs')
print("Jam kerja Lisa =", res.x[0])
print("Jam kerja Muri =", res.x[1])
print("Status:", res.message)
```

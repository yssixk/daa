```python
from pulp import LpProblem, LpVariable, LpMinimize, LpStatus

model = LpProblem("Pemecahan_masalah", sense=LpMinimize)
x = LpVariable("x", lowBound=0)
y = LpVariable("y", lowBound=0)
model += x + y, "Obj"
model += 4 * x + 3 * y == 34, "kendala_2"
model.solve()
print("status:", LpStatus[model.status])
print("x =", x.value())
print("y =", y.value())
```

```python
from pulp import LpProblem, LpVariable, LpMinimize, LpStatus

model = LpProblem("Sistem_Persamaan", sense=LpMinimize)
P = LpVariable("P")
B = LpVariable("B")
model += P + B, "Obj"
model += 3 * P + 4 * B == 11000, "Persamaan_1"
model += P + 7 * B == 15000, "Persamaan_2"
model.solve()
print("status:", LpStatus[model.status])
print("P =", P.value())
print("B =", B.value())
```

```python
# Sistem persamaan:
# Persamaan 1: 3P + 2B = 21000
# Persamaan 2: P + 7B = 15000

from sympy import symbols, Eq, solve

P, B = symbols('P B')

eq1 = Eq(3*P + 2*B, 21000)
eq2 = Eq(P + 7*B, 15000)

solusi = solve((eq1, eq2), (P, B))

harga_pensil = solusi[P]
harga_buku = solusi[B]

total_2P = 2 * harga_pensil
total_6B = 6 * harga_buku

print("Harga pensil (P):", round(harga_pensil, 2))
print("Harga buku (B):", round(harga_buku, 2))
print("Total 2 pensil (2P): Rp", round(total_2P, 2))
print("Total 6 buku (6B): Rp", round(total_6B, 2))
```

```python
def bubbleSort(lst):
    lastElementIndex = len(lst) - 1
    for passNo in range(lastElementIndex, 0, -1):
        for idx in range(passNo):
            if lst[idx] > lst[idx+1]:
                lst[idx], lst[idx+1] = lst[idx+1], lst[idx]
    return lst  
```

```python
def InsertionSort(list):
    lastElementIndex  = len(list) - 1
    for i in range(1, lastElementIndex + 1):
        j = i - 1
        next = list[i]
        while (list[j] > next) and (j >= 0):
            list[j + 1] = list[j]
            j =j-1
        list[j + 1] = next
    return list
```

```python
def selectionSort(list):
    for fill_slot in range(len(list)-1,0,-1):
        Max=0
        for location in range(1,fill_slot+1):
            if list[location]>list[Max]:
                Max = location

        list[fill_slot],list[Max] = list[Max],list[fill_slot]
    return list
```

```python
list = [89, 12, 57, 16, 25]
```

```python
selectionSort(list)
```

```python
def LinearSearch(list, item):
    index = 0
    found = False

    while      index < len(list) and found is False:
        if list [index] == item:
            found = True
        else:
            index = index + 1
    return found
```

```python
list = [12, 33, 11, 99, 22, 55, 90]
print(LinearSearch(list, 12))
print(LinearSearch(list, 100))
```

```python
def BinarySearch(list, item):
    first = 0
    last = len(list)-1
    found = False

    while first <= last and not found:
        midpoint = (first + last)//2
        if list[midpoint] == item:
            found = True
        else:
            if item < list[midpoint]:
                last = midpoint-1
            else:
                first = midpoint+1
    return found
```

```python
list = [12, 33, 11, 99, 22, 55, 90]
sorted_list = bubbleSort(list)
print(BinarySearch(list, 33))
print(BinarySearch(list, 1000))
```

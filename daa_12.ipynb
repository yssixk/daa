```python
import networkx as nx
import matplotlib.pyplot as plt


#Latihan 1 - Ukurlah Centrality
G = nx.Graph()

edges = [
    (5, 1),
    (5, 6),
    (1, 2),
    (6, 3),
    (2, 3),
    (2, 4),
    (3, 4)
]

G.add_edges_from(edges)

print("=" * 60)
print("LATIHAN 1 - UKURLAH CENTRALITY")
print("=" * 60)

# 1. DEGREE CENTRALITY
print("\n1. DEGREE CENTRALITY")
print("-" * 60)
degree_centrality = nx.degree_centrality(G)
for node in sorted(degree_centrality.keys()):
    print(f"Node {node}: {degree_centrality[node]:.4f}")

# 2. BETWEENNESS CENTRALITY
print("\n2. BETWEENNESS CENTRALITY")
print("-" * 60)
betweenness_centrality = nx.betweenness_centrality(G)
for node in sorted(betweenness_centrality.keys()):
    print(f"Node {node}: {betweenness_centrality[node]:.4f}")

# 3. CLOSENESS CENTRALITY
print("\n3. CLOSENESS CENTRALITY")
print("-" * 60)
closeness_centrality = nx.closeness_centrality(G)
for node in sorted(closeness_centrality.keys()):
    print(f"Node {node}: {closeness_centrality[node]:.4f}")

# 4. EIGENVECTOR CENTRALITY
print("\n4. EIGENVECTOR CENTRALITY")
print("-" * 60)
eigenvector_centrality = nx.eigenvector_centrality(G)
for node in sorted(eigenvector_centrality.keys()):
    print(f"Node {node}: {eigenvector_centrality[node]:.4f}")

print("\n" + "=" * 60)


print("\nMembuat visualisasi graf...")

pos = {
    5: (0, 1),      # Top-left
    1: (1, 1),      # Top-right
    6: (0, 0),      # Middle-left
    2: (1, 0),      # Middle-right
    3: (0.5, -0.7), # Bottom-left
    4: (1.5, -0.7)  # Bottom-right
}

# Membuat visualisasi graf
plt.figure(figsize=(10, 8))
nx.draw_networkx_nodes(G, pos, node_color='#5B8FC7', node_size=2000, alpha=0.9)
nx.draw_networkx_labels(G, pos, font_size=16, font_color='white', font_weight='bold')
nx.draw_networkx_edges(G, pos, width=2, alpha=0.6, edge_color='gray')
plt.title("Graf Network - Latihan 1 Ukurlah Centrality", fontsize=14, fontweight='bold', pad=20)
plt.axis('off')
plt.tight_layout()
plt.savefig('network_graph.png', dpi=300, bbox_inches='tight', facecolor='white')
plt.show()
```

```python
import networkx as nx
import matplotlib.pyplot as plt

#Latihan 2 - Hitung Eigenvector Masing-masing Node
G = nx.Graph()
edges = [
    ('Mike', 'Bob'),
    ('Mike', 'Jill'),
    ('Mike', 'Emma'),
    ('Bob', 'Jill'),
    ('Bob', 'John'),
    ('Bob', 'Emma'),
    ('Jill', 'John'),
    ('Jill', 'Emma'),
    ('Jill', 'Shane'),
    ('Jill', 'Leah'),
    ('John', 'Leah'),
    ('Emma', 'Shane'),
    ('Emma', 'Liz'),
    ('Shane', 'Leah'),
    ('Liz', 'Allen'),
    ('Allen', 'Lisa')
]

G.add_edges_from(edges)

print("=" * 60)
print("LATIHAN 2 - HITUNG EIGENVECTOR MASING-MASING NODE")
print("=" * 60)

# Hitung semua ukuran centrality
print("\n1. DEGREE CENTRALITY")
print("-" * 60)
degree_centrality = nx.degree_centrality(G)
for node in sorted(degree_centrality.keys()):
    print(f"{node:10s}: {degree_centrality[node]:.4f}")

print("\n2. BETWEENNESS CENTRALITY")
print("-" * 60)
betweenness_centrality = nx.betweenness_centrality(G)
for node in sorted(betweenness_centrality.keys()):
    print(f"{node:10s}: {betweenness_centrality[node]:.4f}")

print("\n3. CLOSENESS CENTRALITY")
print("-" * 60)
closeness_centrality = nx.closeness_centrality(G)
for node in sorted(closeness_centrality.keys()):
    print(f"{node:10s}: {closeness_centrality[node]:.4f}")

print("\n4. EIGENVECTOR CENTRALITY")
print("-" * 60)
eigenvector_centrality = nx.eigenvector_centrality(G)
for node in sorted(eigenvector_centrality.keys()):
    print(f"{node:10s}: {eigenvector_centrality[node]:.4f}")

# Mencari node dengan eigenvector centrality tertinggi
max_eigen_node = max(eigenvector_centrality, key=eigenvector_centrality.get)
max_eigen_value = eigenvector_centrality[max_eigen_node]

print("\n" + "=" * 60)
print(f"Node dengan Eigenvector Centrality Tertinggi: {max_eigen_node}")
print(f"Nilai: {max_eigen_value:.4f}")
print("=" * 60)

pos = nx.spring_layout(G, seed=42, k=0.5, iterations=50)

plt.figure(figsize=(12, 9))

#nodes
nx.draw_networkx_nodes(G, pos, node_color='#5B8FC7', node_size=1500, alpha=0.9)

#labels
nx.draw_networkx_labels(G, pos, font_size=11, font_color='white', font_weight='bold')

#edges
nx.draw_networkx_edges(G, pos, width=2, alpha=0.5, edge_color='#6495ED')

plt.title("Graf Network - Latihan 2\n(Hitung Eigenvector Masing-masing Node)", 
          fontsize=14, fontweight='bold', pad=20)
plt.axis('off')
plt.tight_layout()
plt.savefig('network_graph_latihan2.png', dpi=300, bbox_inches='tight', facecolor='white')
plt.show()

```

```python
import networkx as nx
import matplotlib.pyplot as plt
import urllib.request
import gzip
import os

print("=" * 60)
print("LATIHAN 4 - FACEBOOK NETWORK DARI STANFORD SNAP")
print("=" * 60)

#Link untuk download data
url = "https://snap.stanford.edu/data/facebook_combined.txt.gz"
filename = "facebook_combined.txt.gz"
txt_filename = "facebook_combined.txt"

print("\nMengunduh data dari Stanford SNAP...")
print(f"URL: {url}")

#Download file if not exists
if not os.path.exists(txt_filename):
    if not os.path.exists(filename):
        print("Mengunduh file...")
        urllib.request.urlretrieve(url, filename)
        print("Download selesai!")
    
    #Extract file
    print("Mengekstrak file...")
    with gzip.open(filename, 'rb') as f_in:
        with open(txt_filename, 'wb') as f_out:
            f_out.write(f_in.read())
    print("Ekstraksi selesai!")
else:
    print("File sudah ada, menggunakan file yang ada...")

print("\nMemuat data ke NetworkX...")
#Read edge list
G = nx.read_edgelist(txt_filename, nodetype=int)

print("\n" + "=" * 60)
print("STATISTIK DATASET")
print("=" * 60)
print(f"Jumlah Nodes: {G.number_of_nodes()}")
print(f"Jumlah Edges: {G.number_of_edges()}")
print(f"Rata-rata Degree: {sum(dict(G.degree()).values()) / G.number_of_nodes():.2f}")

# Hitung centrality measures
print("\n" + "=" * 60)
print("MENGHITUNG CENTRALITY MEASURES")
print("=" * 60)

print("\nMenghitung Degree Centrality...")
degree_centrality = nx.degree_centrality(G)

print("Menghitung Betweenness Centrality...")
betweenness_centrality = nx.betweenness_centrality(G, k=100)  # Sampling untuk efisiensi

print("Menghitung Closeness Centrality...")
closeness_centrality = nx.closeness_centrality(G)

print("Menghitung Eigenvector Centrality...")
eigenvector_centrality = nx.eigenvector_centrality(G, max_iter=100)

# Top 10 nodes untuk setiap centrality
print("\n" + "=" * 60)
print("TOP 10 NODES - DEGREE CENTRALITY")
print("-" * 60)
top_degree = sorted(degree_centrality.items(), key=lambda x: x[1], reverse=True)[:10]
for i, (node, value) in enumerate(top_degree, 1):
    print(f"{i:2d}. Node {node:4d}: {value:.4f}")

print("\n" + "=" * 60)
print("TOP 10 NODES - BETWEENNESS CENTRALITY")
print("-" * 60)
top_betweenness = sorted(betweenness_centrality.items(), key=lambda x: x[1], reverse=True)[:10]
for i, (node, value) in enumerate(top_betweenness, 1):
    print(f"{i:2d}. Node {node:4d}: {value:.4f}")

print("\n" + "=" * 60)
print("TOP 10 NODES - CLOSENESS CENTRALITY")
print("-" * 60)
top_closeness = sorted(closeness_centrality.items(), key=lambda x: x[1], reverse=True)[:10]
for i, (node, value) in enumerate(top_closeness, 1):
    print(f"{i:2d}. Node {node:4d}: {value:.4f}")

print("\n" + "=" * 60)
print("TOP 10 NODES - EIGENVECTOR CENTRALITY")
print("-" * 60)
top_eigenvector = sorted(eigenvector_centrality.items(), key=lambda x: x[1], reverse=True)[:10]
for i, (node, value) in enumerate(top_eigenvector, 1):
    print(f"{i:2d}. Node {node:4d}: {value:.4f}")

print("\n" + "=" * 60)
print("VISUALISASI GRAF")
print("=" * 60)
print("Catatan: Karena graph memiliki 4039 nodes, visualisasi penuh")
print("akan sangat padat. Membuat visualisasi dengan layout spring...")
plt.figure(figsize=(14, 12))
pos = nx.spring_layout(G, k=0.15, iterations=20, seed=42)
node_sizes = [degree_centrality[node] * 3000 for node in G.nodes()]

nx.draw_networkx_nodes(G, pos, 
                       node_color='#4267B2',  # Facebook blue color
                       node_size=node_sizes,
                       alpha=0.6,
                       edgecolors='#1877F2',
                       linewidths=0.5)

nx.draw_networkx_edges(G, pos, 
                       width=0.1,
                       alpha=0.2,
                       edge_color='gray')

plt.title("Facebook Network - Stanford SNAP Dataset\n(4039 nodes, 88234 edges)", 
          fontsize=16, fontweight='bold', pad=20)
plt.axis('off')
plt.tight_layout()
plt.savefig('facebook_network_graph.png', dpi=300, bbox_inches='tight', facecolor='white')
plt.show()

print("\nGraf berhasil disimpan sebagai 'facebook_network_graph.png'")
print("Selesai!")
```

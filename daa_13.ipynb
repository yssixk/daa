```python
grafik = {
    'A': ['B', 'C', 'D'],
    'B': ['D', 'E'],
    'C': ['F', 'E'],
    'D': ['B', 'A'],
    'E': ['F', 'C'],
    'F': ['E', 'C']}
```

```python
def bfs(grafik, start):
    visited = []  
    queue = []    
    queue.append(start)
    visited.append(start)
    
    while queue:
        current = queue.pop(0)
        for neighbor in grafik[current]:
            if neighbor not in visited:
                visited.append(neighbor)
                queue.append(neighbor)
    return visited
```

```python
bfs(grafik, 'B')
```

```python
def dfs(graph, start):
    visited = []
    stack = []
    stack.append(start)
    while stack:
        current = stack.pop()
        if current not in visited:
            visited.append(current)
            for neighbor in reversed(graph[current]):
                if neighbor not in visited:
                    stack.append(neighbor)
    return visited
grafik = {
    'A': ['B', 'C', 'D'],
    'B': ['D', 'E'],
    'C': ['F', 'E'],
    'D': ['B', 'A'],
    'E': ['F', 'C'],
    'F': ['E', 'C']}
```

```python
while dfs(grafik, 'A'):
    print(dfs(grafik, 'A'))
    break
while dfs(grafik, 'D'):
    print(dfs(grafik, 'D'))
    break
```
